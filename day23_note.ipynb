**Task 4: Data Cleaning**

```python
data.isnull().sum()

# dropping all missing values rows
data.dropna(inplace=True)

# data.fillna()
data.isnull().sum()
```

```python
# dealing with duplicates rows
data.duplicated().sum()

# deleting all duplicated rows
# data[data.duplicated()]

data.drop_duplicates(inplace=True)
data.duplicated().sum()
```

*** ALL THE CARTS IN GOOD PICTURING ***

```python
# ══════════════════════════════════════════════════════════════════
#  DROP THIS INTO YOUR NOTEBOOK — replaces all chart cells
#  Works with:  new_york_listings_2024.csv
#  Run after your data cleaning step (df = data[data['price'] < 1000])
# ══════════════════════════════════════════════════════════════════

import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns
import numpy as np

# ─────────────────────────────────────────────
#  AIRBNB COLOR PALETTE
# ─────────────────────────────────────────────
CORAL  = '#FF5A5F'
TEAL   = '#00A699'
ORANGE = '#FC642D'
DARK   = '#484848'
GRAY   = '#767676'
WHITE  = '#FFFFFF'
NAVY   = '#3D5A80'
PALETTE = [CORAL, TEAL, ORANGE, NAVY, '#8338EC']

# ─────────────────────────────────────────────
#  GLOBAL STYLE  (run once at the top)
# ─────────────────────────────────────────────
plt.rcParams.update({
    'figure.facecolor' : WHITE,
    'axes.facecolor'   : WHITE,
    'axes.spines.top'  : False,
    'axes.spines.right': False,
    'axes.spines.left' : False,
    'axes.spines.bottom': True,
    'axes.edgecolor'   : '#DDDDDD',
    'axes.labelcolor'  : DARK,
    'axes.labelsize'   : 12,
    'axes.titlesize'   : 16,
    'axes.titleweight' : 'bold',
    'axes.titlecolor'  : DARK,
    'axes.titlepad'    : 14,
    'xtick.color'      : GRAY,
    'ytick.color'      : GRAY,
    'xtick.labelsize'  : 10,
    'ytick.labelsize'  : 10,
    'grid.color'       : '#F0F0F0',
    'grid.linewidth'   : 0.8,
    'legend.frameon'   : False,
    'legend.fontsize'  : 10,
})

def brand(fig):
    fig.text(0.5, 0.005, 'New York Airbnb 2024  •  EDA Project',
             ha='center', va='bottom', fontsize=8.5, color=GRAY, style='italic')

# ──────────────────────────────────────────────────────────────────
#  CHART 1 — Price Distribution
# ──────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(11, 5.5))

sns.histplot(df, x='price', bins=80, color=CORAL, alpha=0.75,
             kde=True, line_kws={'color': DARK, 'lw': 2.2, 'ls': '--'}, ax=ax)

med, mean = df['price'].median(), df['price'].mean()
ax.axvline(med,  color=TEAL,   lw=2, ls='--', label=f'Median  ${med:.0f}')
ax.axvline(mean, color=ORANGE, lw=2, ls='--', label=f'Mean    ${mean:.0f}')

ax.set_title('Price Distribution of NYC Airbnb Listings')
ax.set_xlabel('Nightly Price (USD)'); ax.set_ylabel('Number of Listings')
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'${int(x):,}'))
ax.yaxis.grid(True, alpha=0.6); ax.set_axisbelow(True)
ax.legend(loc='upper right', fontsize=11)
ax.annotate('Most listings priced\nbetween $50 – $300',
            xy=(200, ax.get_ylim()[1] * 0.72),
            fontsize=9.5, color=DARK,
            bbox=dict(boxstyle='round,pad=0.4', fc='#FFF3F3', ec=CORAL, alpha=0.9))

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart1_price_distribution.png', dpi=180, bbox_inches='tight')
plt.show()

# ──────────────────────────────────────────────────────────────────
#  CHART 2 — Price by Borough & Room Type
# ──────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(13, 6))
b_order = ['Manhattan', 'Brooklyn', 'Queens', 'Bronx', 'Staten Island']

sns.barplot(df, x='neighbourhood_group', y='price', hue='room_type',
            order=b_order, palette=PALETTE[:4],
            edgecolor='white', linewidth=0.8, ax=ax)

ax.set_title('Average Price by Borough & Room Type')
ax.set_xlabel('Borough'); ax.set_ylabel('Avg Nightly Price (USD)')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda y, _: f'${int(y)}'))
ax.yaxis.grid(True, alpha=0.6); ax.set_axisbelow(True)
ax.legend(title='Room Type', bbox_to_anchor=(1.01, 1), loc='upper left', title_fontsize=10)

for p in ax.patches:      # value labels on bars
    h = p.get_height()
    if h > 0:
        ax.text(p.get_x() + p.get_width() / 2, h + 2, f'${h:.0f}',
                ha='center', va='bottom', fontsize=8, color=DARK)

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart2_price_by_borough.png', dpi=180, bbox_inches='tight')
plt.show()

# ──────────────────────────────────────────────────────────────────
#  CHART 3 — Availability Distribution
# ──────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(11, 5.5))

sns.histplot(df, x='availability_365', bins=60, color=TEAL, alpha=0.75,
             kde=True, line_kws={'color': DARK, 'lw': 2.2, 'ls': '--'}, ax=ax)

ax.axvspan(300, 365, alpha=0.10, color=TEAL, label='High availability (300+ days)')
ax.set_title('Listing Availability Distribution (Days / Year)')
ax.set_xlabel('Available Days (out of 365)'); ax.set_ylabel('Number of Listings')
ax.yaxis.grid(True, alpha=0.6); ax.set_axisbelow(True)
ax.legend(fontsize=10)

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart3_availability.png', dpi=180, bbox_inches='tight')
plt.show()

# ──────────────────────────────────────────────────────────────────
#  CHART 4 — Reviews vs Price Scatter
# ──────────────────────────────────────────────────────────────────
boroughs = ['Manhattan', 'Brooklyn', 'Queens', 'Bronx', 'Staten Island']
b_colors = dict(zip(boroughs, PALETTE))

fig, ax = plt.subplots(figsize=(11, 6.5))
sample = df.sample(min(4000, len(df)), random_state=7)

for b in boroughs:
    sub = sample[sample['neighbourhood_group'] == b]
    ax.scatter(sub['number_of_reviews'], sub['price'],
               c=b_colors[b], alpha=0.38, s=20, label=b, edgecolors='none')

ax.set_title('Price vs. Number of Reviews by Borough')
ax.set_xlabel('Number of Reviews'); ax.set_ylabel('Nightly Price (USD)')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda y, _: f'${int(y)}'))
ax.yaxis.grid(True, alpha=0.6); ax.xaxis.grid(True, alpha=0.4)
ax.set_axisbelow(True)
ax.legend(title='Borough', bbox_to_anchor=(1.01, 1), loc='upper left',
          title_fontsize=10, markerscale=2)

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart4_reviews_vs_price.png', dpi=180, bbox_inches='tight')
plt.show()

# ──────────────────────────────────────────────────────────────────
#  CHART 5 — Geographical Map
# ──────────────────────────────────────────────────────────────────
r_colors = {'Entire home/apt': CORAL, 'Private room': TEAL,
            'Shared room': ORANGE, 'Hotel room': NAVY}

fig, ax = plt.subplots(figsize=(11, 8))
ax.set_facecolor('#EAF2F8')
sample_g = df.sample(min(6000, len(df)), random_state=3)

for rt, col in r_colors.items():
    sub = sample_g[sample_g['room_type'] == rt]
    ax.scatter(sub['longitude'], sub['latitude'],
               c=col, alpha=0.28, s=9, label=rt, edgecolors='none')

ax.set_title('Geographical Distribution of NYC Airbnb Listings')
ax.set_xlabel('Longitude'); ax.set_ylabel('Latitude')
ax.grid(True, alpha=0.3, color='white', lw=0.8); ax.set_axisbelow(True)
ax.legend(title='Room Type', markerscale=3, bbox_to_anchor=(1.01, 1),
          loc='upper left', title_fontsize=10)

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart5_geo_map.png', dpi=180, bbox_inches='tight')
plt.show()

# ──────────────────────────────────────────────────────────────────
#  CHART 6 — Correlation Heatmap
# ──────────────────────────────────────────────────────────────────
num_cols = ['price', 'minimum_nights', 'number_of_reviews',
            'reviews_per_month', 'availability_365', 'beds']
labels   = ['Price', 'Min Nights', '# Reviews', 'Reviews/Mo', 'Availability', 'Beds']
corr = df[num_cols].corr()
corr.index = labels; corr.columns = labels

fig, ax = plt.subplots(figsize=(9, 7.5))
mask_upper = np.triu(np.ones_like(corr, dtype=bool), k=1)
cmap = sns.diverging_palette(10, 175, s=85, l=50, as_cmap=True)

sns.heatmap(corr, mask=mask_upper, annot=True, fmt='.2f',
            cmap=cmap, center=0, square=True,
            linewidths=1.2, linecolor='white',
            annot_kws={'size': 10, 'weight': 'bold'},
            cbar_kws={'shrink': 0.75, 'pad': 0.02}, ax=ax)

ax.set_title('Feature Correlation Matrix')
ax.tick_params(axis='x', rotation=30); ax.tick_params(axis='y', rotation=0)

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart6_heatmap.png', dpi=180, bbox_inches='tight')
plt.show()

# ──────────────────────────────────────────────────────────────────
#  CHART 7 — Room Type Donut
# ──────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 7))
counts = df['room_type'].value_counts()

wedges, _, autotexts = ax.pie(
    counts, labels=None, autopct='%1.1f%%', startangle=90,
    colors=PALETTE[:len(counts)], pctdistance=0.75,
    wedgeprops={'width': 0.58, 'edgecolor': 'white', 'linewidth': 2.8}
)
for at in autotexts:
    at.set_fontsize(12); at.set_fontweight('bold'); at.set_color('white')

ax.text(0, 0, f'{len(df):,}\nListings', ha='center', va='center',
        fontsize=13, fontweight='bold', color=DARK)

ax.legend(wedges, counts.index, title='Room Type',
          loc='lower center', bbox_to_anchor=(0.5, -0.06), ncol=2, fontsize=10)
ax.set_title('Room Type Distribution')

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart7_room_type_donut.png', dpi=180, bbox_inches='tight')
plt.show()

# ──────────────────────────────────────────────────────────────────
#  CHART 8 — Price Boxplot by Borough
# ──────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(12, 6))
b_order2 = df.groupby('neighbourhood_group')['price'].median() \
              .sort_values(ascending=False).index.tolist()

sns.boxplot(df, x='neighbourhood_group', y='price', order=b_order2,
            palette=PALETTE[:len(b_order2)],
            fliersize=2.5, flierprops={'alpha': 0.25, 'markeredgecolor': 'none'},
            linewidth=1.3, width=0.55, ax=ax)

for i, b in enumerate(b_order2):
    med_val = df[df['neighbourhood_group'] == b]['price'].median()
    ax.text(i, med_val + 8, f'${med_val:.0f}', ha='center', va='bottom',
            fontsize=9.5, fontweight='bold', color=WHITE,
            bbox=dict(boxstyle='round,pad=0.25', fc=PALETTE[i], alpha=0.85))

ax.set_title('Price Distribution by Borough  (Sorted by Median)')
ax.set_xlabel('Borough'); ax.set_ylabel('Nightly Price (USD)')
ax.yaxis.set_major_formatter(mticker.FuncFormatter(lambda y, _: f'${int(y)}'))
ax.yaxis.grid(True, alpha=0.6); ax.set_axisbelow(True)

plt.tight_layout(rect=[0, 0.03, 1, 1])
brand(fig); plt.savefig('chart8_price_boxplot.png', dpi=180, bbox_inches='tight')
plt.show()
```
